```python
import fairlearn
import pandas as pd
from fairlearn import datasets
```

```python
from fairlearn import datasets

# Load a sample dataset from fairlearn
X, y, sensitive_features = datasets.fetch_acs_income(as_frame=True)

# Combine features and target into a single DataFrame for easier processing
df_acs_income = pd.concat([X, y, sensitive_features], axis=1)

print("Info for the acs_income dataset:")
df_acs_income.info()

print("\nFirst 5 rows of the acs_income dataset:")
display(df_acs_income.head())
```

```python
# Load the diabetes_hospital dataset
X_diabetes, y_diabetes = datasets.fetch_diabetes_hospital(as_frame=True)

# Combine features and target into a single DataFrame
df_diabetes = pd.concat([X_diabetes, y_diabetes], axis=1)

print("Info for the diabetes_hospital dataset:")
df_diabetes.info()

print("\nFirst 5 rows of the diabetes_hospital dataset:")
display(df_diabetes.head())
```

```python
# Load the credit_card dataset
X_credit, y_credit, sensitive_features_credit = datasets.fetch_credit_card(as_frame=True)

# Combine features and target into a single DataFrame
df_credit = pd.concat([X_credit, y_credit, sensitive_features_credit], axis=1)

print(f"Full dataset size: {len(df_credit)} rows\n")

print("Info for the credit_card dataset:")
df_credit.info()

print("\nFirst 5 rows of the credit_card dataset:")
display(df_credit.head())
```
